## 1단계: 실행 환경 준비하기

Kaggle이 빌려준 컴퓨터의 메모리를 청소하고, AI 모델이 잘 돌아가도록 설정값을 맞춥니다.
**몇 초면 끝납니다.**

````python
import os, gc, psutil

print('=== Kaggle 실행 환경을 준비합니다 (GPU T4 x2) ===')

# 지금 이 컴퓨터에 메모리(RAM)가 얼마나 있는지 확인만 해봅니다.
mem = psutil.virtual_memory()
print(f'메모리: 전체 {mem.total / 1024**3:.1f} GB / 지금 쓸 수 있는 양 {mem.available / 1024**3:.1f} GB')

# 안 쓰는 캐시를 비워서 메모리를 최대한 확보합니다.
# (실패해도 상관없어서 오류 메시지는 숨겨둡니다)
os.system('echo 3 | sudo tee /proc/sys/vm/drop_caches > /dev/null 2>&1')
os.system('echo 1 | sudo tee /proc/sys/vm/overcommit_memory > /dev/null 2>&1')
gc.collect()

# 아래 3줄은 AI 모델이 GPU 메모리를 아껴 쓰도록 알려주는 설정입니다.
# 이게 없으면 "CUDA out of memory" 오류가 나기 쉽습니다.
os.environ['PYTORCH_CUDA_ALLOC_CONF'] = 'expandable_segments:True,garbage_collection_threshold:0.6'
os.environ['MALLOC_TRIM_THRESHOLD_'] = '0'
os.environ['TOKENIZERS_PARALLELISM'] = 'false'

# ── 다운로드 방식 선택 (USE_XET) ────────────────────────────
# 허깅페이스에서 모델을 받는 방식은 두 가지입니다.
#   True  = Xet 방식 (원본 노트북과 동일하지만, 멈추는 등 불안정할 때가 있음)
#   False = 예전 HTTP 방식 (이 노트북의 기본값 — 안정적, 403 오류는 자동 재시도)
# 어느 쪽이 빠른지는 그날그날 허깅페이스 서버 사정에 따라 다릅니다.
# 👉 3단계 다운로드가 1%에서 멈춘 것처럼 보이면,
#    아래 값을 반대로 바꾸고 Run → Restart & Run All 하세요.
# ※ 반드시 huggingface_hub 을 부르기 전에 정해야 해서 여기(1단계)에 있습니다.
USE_XET = True
if USE_XET:
    # True 일 때는 원본 노트북과 '완전히 같은 상태'가 되도록 아무 설정도 남기지 않습니다.
    os.environ.pop('HF_HUB_DISABLE_XET', None)
else:
    os.environ['HF_HUB_DISABLE_XET'] = '1'

# [행글라이터] 네트워크가 잠깐 끊겨도 쉽게 포기하지 않도록 대기 시간을 넉넉히 줍니다.
# (기본값은 10초입니다. 큰 파일을 오래 받는 동안 한 번 끊기는 건 흔한 일입니다)
os.environ['HF_HUB_DOWNLOAD_TIMEOUT'] = '60'


print('✅ 1단계 완료! 환경 준비가 끝났습니다.')
print('   (다운로드 방식: ' + ('Xet — 원본 노트북과 동일' if USE_XET else '예전 HTTP 방식') + ')')
````

⏱️ **3~5분 정도 걸립니다.** 글자가 주르륵 올라가는 건 정상입니다.

> ### 😱 빨간 글씨가 잔뜩 나와도 정상입니다!
>
> 설치가 끝날 무렵 아래 같은 글이 **수십 줄** 쏟아집니다.
>
> ```
> ERROR: pip's dependency resolver does not currently take into account all the packages...
> bigframes 2.39.0 requires google-cloud-bigquery-storage..., which is not installed.

> **이건 실패가 아닙니다.** `ERROR`라고 적혀 있지만, 실제로는 pip이

> **Kaggle이 기본으로 깔아둔, 우리가 쓰지도 않는 패키지들**입니다. 그냥 넘어가시면 됩니다.

> ### 🔄 커널을 다시 시작해야 하나요? → **아니요!**
>
> 원본 노트북에는 *"이 셀을 실행한 뒤 커널을 재시작하고 다시 실행하세요"* 라는 안내가 있었습니다.
> **이 한국어판에는 그 과정이 필요 없습니다.**
>
> 원본이 재시작을 요구했던 이유는 **P100 그래픽카드를 쓰려고 PyTorch를 옛날 버전(2.3.1)으로
> 되돌렸기 때문**입니다. 우리는 **T4 를 쓰기 때문에 그럴 필요가 없어서 그 과정을 아예 없앴습니다.**
>
> 게다가 이 노트북의 셀들은 파이썬 기본 기능만 쓰고, 진짜 무거운 것들(torch 등)은
> 5단계에서 **완전히 새로운 프로그램으로 따로 실행**됩니다.
> 새로 시작하는 프로그램이라 방금 설치한 버전을 알아서 읽어갑니다.

````python
import os, subprocess

# ── (1) GPU 확인 ────────────────────────────────────────────
# 이 노트북은 T4 전용입니다. P100 을 고르면 훨씬 느리고 오류도 납니다.
print('=== GPU 확인 ===')
try:
    out = subprocess.run(['nvidia-smi', '--query-gpu=name,memory.total',
                          '--format=csv,noheader'], capture_output=True, text=True, check=True).stdout.strip()
    print(out)
    if 'P100' in out:
        print('\n⛔ P100 이 선택되어 있습니다! 이 노트북은 T4 전용입니다.')
        print('   → Settings → Accelerator → GPU T4 x2 로 바꾼 뒤 Restart & Run All 하세요.')
        print('   (T4 에는 AI 계산 전용 회로가 있어서 P100 보다 3배 이상 빠릅니다)')
    elif 'T4' in out:
        print('✅ T4 확인! 잘 선택하셨습니다.')
    else:
        print('ℹ️ T4 가 아닌 GPU 입니다. 동작할 수도 있지만 이 노트북은 T4 기준으로 맞춰져 있습니다.')
except Exception:
    print('⛔ GPU 가 없습니다! (CPU 세션에서는 이 노트북이 동작하지 않습니다)')
    print('   새 노트북 파일을 올리면 Kaggle 설정이 초기화됩니다!')
    print('   → 오른쪽 Settings → Accelerator → GPU T4 x2 선택')
    print('   → 같은 곳에서 Internet 도 On 인지 확인')
    print('   → 그 뒤 Run → Restart & Run All 로 처음부터 다시 실행하세요.')
    raise SystemExit('GPU 미설정 — 위 안내대로 설정한 뒤 처음부터 다시 실행하세요.')

# ── (2) Wan2GP 프로그램 내려받기 ────────────────────────────
# 이미 받아둔 적이 있으면 건너뜁니다.
# (이 확인이 없으면 셀을 두 번 실행할 때 빨간 오류가 납니다)
print('\n=== Wan2GP 내려받기 ===')
if os.path.isdir('Wan2GP'):
    print('✅ Wan2GP 폴더가 이미 있습니다. 내려받기를 건너뜁니다.')
else:
    !git clone https://github.com/DeepBeepMeep/Wan2GP.git
    # [행글라이터] 검증된 버전으로 고정합니다.
    # 고정하지 않으면 Wan2GP가 업데이트되는 순간 이 노트북이 갑자기 안 돌아갑니다.
    # (영상은 계속 남는데 노트북만 깨지는 사고를 막기 위한 장치입니다)
    !cd Wan2GP && git checkout -q 5582327dc25e45fec6cda0f27144d4dcf7ed104b
    print('✅ Wan2GP 버전 고정: 5582327 (2026-07-11 v12.3 — 행글라이터 검증본)')

# ── (3) 필요한 파이썬 패키지 설치 ───────────────────────────
# ※ 원본에는 여기서 PyTorch 를 2.3.1 로 되돌리는 줄이 있었습니다.
#    그건 P100(sm_60) 이 최신 PyTorch 에서 잘려나갔기 때문인데,
#    우리가 쓰는 T4(sm_75) 는 최신 PyTorch 를 그대로 쓸 수 있어서 지웠습니다.
#    덕분에 '커널 재시작' 과정도 필요 없어졌습니다.
print('\n=== 패키지 설치 (3~5분) ===')
!pip install --timeout 120 --retries 5 -q -r Wan2GP/requirements.txt
!pip install --timeout 120 --retries 5 -q mmgp gradio gguf soundfile

# ── (4) 실제로 설치된 버전 확인 ─────────────────────────────
# 나중에 5단계에서 오류가 나면 여기 찍힌 버전을 보고 원인을 찾을 수 있습니다.
print('\n=== 설치된 버전 (문제 생기면 이 부분을 캡처해 주세요) ===')
import importlib.metadata as _md
for pkg in ['torch', 'numpy', 'gradio', 'transformers', 'huggingface_hub', 'mmgp', 'gguf']:
    try:
        print(f'  {pkg:18s} {_md.version(pkg)}')
    except Exception:
        print(f'  {pkg:18s} (설치 안 됨)')

print('\n✅ 2단계 완료! 설치가 끝났습니다.')
````

| VAE / 업스케일러 등 | 2.9 GB | 결과를 실제 영상으로 바꾸는 부분 |

**이 모델들을 만든 곳** (공개해 주셔서 감사합니다)

| 실행 엔진 Wan2GP | [DeepBeepMeep](https://github.com/DeepBeepMeep/Wan2GP) |

⏱️ **보통 5분 안팎**이면 끝나지만, 그날 사정에 따라 더 걸리기도 합니다.
진행 막대가 움직이면 잘 되고 있는 겁니다.

> 💡 Kaggle은 저장 공간이 나뉘어 있습니다. 큰 파일은 여유 공간이 많은 `/kaggle/tmp`에 받고,
> 프로그램이 찾는 위치에는 **바로가기(심볼릭 링크)**만 걸어둡니다.
> `/kaggle/working`은 20GB 제한이라 이렇게 하지 않으면 용량이 터집니다.

> ### 🔎 다운로드가 1%에서 멈춘 것 같다면
> 이 셀 맨 위에 지금 **어떤 방식으로 받는지** 표시됩니다 (`Xet` 또는 `예전 HTTP`).
> 두 방식 중 어느 쪽이 빠른지는 **그날그날 허깅페이스 서버 사정에 따라 다릅니다.**
>
> 몇 분을 기다려도 진행률이 안 움직이면:
> **1단계 셀의 `USE_XET` 값을 반대로 바꾸고 (True ↔ False) Run → Restart & Run All** 하세요.
> 받다 만 파일은 남아 있어서 이어받기 때문에 처음부터 다시 받지 않습니다.
>
> ※ 참고: Xet 방식은 진행 막대가 이상하게 보일 수 있습니다 (낮은 숫자에 머물다 한 번에 점프).

> ### ⚠️ "실패 — 1/5회" 같은 글이 나와도 놀라지 마세요
>
> **HTTP 방식**으로 받을 때는 **유효기간이 붙은 서명된 임시 주소**로 파일을 받아오는데,
> 이 서명이 허깅페이스 서버 사정으로 가끔 거부당합니다 (`403 Forbidden`, `SignatureError`).
>
> 다행히 **다시 요청하면 서명 주소를 새로 발급받기 때문에 대부분 두세 번 안에 성공합니다.**
> 그래서 이 셀은 실패하면 **자동으로 5번까지 다시 시도**합니다.
> 아래처럼 나오는 건 고장이 아니라 정상 동작입니다.
>
> ```
> ⚠️ Gemma 인코더 실패 — 1/5회 (HTTP 403)
>    5초 뒤 새 주소를 받아 다시 시도합니다... (정상적인 재시도입니다)
> ```
>
> **받다 만 파일은 남아 있어서, 이 셀만 다시 실행하면 이어받습니다.** 처음부터 다시 받지 않습니다.

````python
import os

os.environ['HF_HOME'] = '/kaggle/tmp/hf_home'          # 임시 캐시는 여유 공간이 많은 곳에
os.environ['HF_HUB_DISABLE_SYMLINKS_WARNING'] = '1'

import shutil, time
from huggingface_hub import hf_hub_download

# 지금 어떤 방식으로 받는지 알려줍니다.
# 👉 다운로드가 멈춘 것 같으면 1단계의 USE_XET 값을 반대로 바꾸고 다시 실행하세요.
try:
    from huggingface_hub import constants as _hf_const
    _xet_off = getattr(_hf_const, 'HF_HUB_DISABLE_XET', True)
    # hf_xet 이 '설치돼 있는지'만 확인합니다 (불러오지는 않아서 원본 동작에 영향 없음)
    import importlib.util
    _xet_available = importlib.util.find_spec('hf_xet') is not None

    if _xet_off or not _xet_available:
        print('ℹ️ 다운로드 방식: 예전 HTTP 방식')
    else:
        print('ℹ️ 다운로드 방식: Xet (원본 노트북과 동일한 기본값)')
    print('   → 다운로드가 1%에서 멈춘 것처럼 보이면: 1단계 셀의 USE_XET 값을 반대로 바꾸고')
    print('     Run → Restart & Run All 하세요. 어느 쪽이 빠른지는 그날 서버 사정에 따라 다릅니다.')
except Exception:
    pass
print()

# 모델 파일을 가져올 인터넷 주소(허깅페이스 저장소)입니다.
REPO = 'Abiray/LTX-2.3-22B-DISTILLED-1.1-GGUF'   # 영상 만드는 두뇌 (GGUF 로 압축된 버전)
COMPANION_REPO = 'DeepBeepMeep/LTX-2'            # 나머지 부품들

MODEL_DIR = 'Wan2GP/models'      # 프로그램이 모델을 찾는 위치 (/kaggle/working — 20GB 제한)
TMP_DIR = '/kaggle/tmp/models'   # 용량 여유가 많은 임시 저장 위치
os.makedirs(MODEL_DIR, exist_ok=True)
os.makedirs(TMP_DIR, exist_ok=True)


def safe_symlink(src, dst):
    """바로가기(심볼릭 링크)를 안전하게 만듭니다.

    셀을 두 번 실행하거나 다운로드가 중간에 끊기면
    껍데기만 남은 바로가기가 생겨서 오류가 납니다.
    그런 경우 낡은 바로가기를 지우고 새로 만듭니다.
    """
    if os.path.islink(dst):
        if os.path.exists(dst):
            return False        # 멀쩡한 바로가기가 이미 있음 → 그대로 둠
        os.unlink(dst)          # 끊어진 바로가기 → 지우고 새로 만듦
    elif os.path.exists(dst):
        return False            # 진짜 파일이 이미 있음 → 그대로 둠
    os.symlink(src, dst)
    return True


def download(what, tries=5, **kw):
    """모델 파일을 내려받되, 실패하면 알아서 다시 시도합니다.

    ── 이 함수가 왜 필요한가 ──
    허깅페이스에서 큰 파일을 받다 보면 일시적인 오류가 생길 수 있습니다.
    특히 HTTP 방식은 '유효기간이 붙은 서명된 임시 주소'로 파일을 받는데,
    이 서명이 서버 사정으로 가끔 거부됩니다(403 Forbidden / SignatureError).

    huggingface_hub 에도 자동 재시도가 들어 있지만,
    재시도 대상이 (408, 429, 500, 502, 503, 504) 뿐이라 403 은 그냥 통과해 버립니다.
    그래서 여기서 한 겹 더 감싸줍니다.

    다시 요청하면 서명 주소를 새로 발급받기 때문에 대부분 두세 번 안에 성공합니다.
    45GB 를 25분 넘게 받는 동안 네트워크가 한 번 끊기는 것도 흔한 일이라,
    이렇게 해두면 처음부터 다시 받을 필요가 없습니다.
    """
    for attempt in range(1, tries + 1):
        try:
            return hf_hub_download(**kw)
        except Exception as e:
            # 파일 이름이 틀렸거나 저장소가 없는 경우는 몇 번을 해도 안 되니 바로 포기합니다.
            code = getattr(getattr(e, 'response', None), 'status_code', None)
            if code == 404:
                print(f'  ⛔ {what}: 파일을 찾을 수 없습니다 (404). 노트북이 낡았을 수 있습니다.')
                raise

            if attempt == tries:
                print(f'\n  ⛔ {what}: {tries}번 시도했지만 모두 실패했습니다.')
                print('     ── 이렇게 해보세요 ──')
                print('     1) 잠시(5~10분) 뒤 이 셀만 다시 실행하세요.')
                print('        받다 만 파일은 남아 있어서 이어받기 때문에 처음부터 다시 받지 않습니다.')
                print('     2) 1단계 셀의 USE_XET 값을 반대로 바꾸고 (True ↔ False)')
                print('        Run → Restart & Run All 로 다른 다운로드 방식을 시도해 보세요.')
                print('     3) 그래도 안 되면 Stop Session 후 새 세션에서 다시 실행하세요.')
                print('        (새 컴퓨터를 배정받으면 다른 서버로 연결되어 뚫리는 경우가 많습니다)')
                raise

            # [행글라이터] 지수 백오프: 5·10·15·20초(총 50초)로는
            # 허깅페이스가 속도 제한(429)을 걸었을 때 못 버팁니다.
            wait = min(10 * (2 ** (attempt - 1)), 160)
            hint = f' (HTTP {code})' if code else ''
            print(f'  ⚠️ {what} 실패 — {attempt}/{tries}회{hint}')
            print(f'     {wait}초 뒤 새 주소를 받아 다시 시도합니다... (정상적인 재시도입니다)')
            time.sleep(wait)


def fetch_and_link(filename, repo, label, link_name=None):
    """/kaggle/tmp 에 받은 뒤, 프로그램이 찾는 위치에 바로가기를 걸어줍니다."""
    dest = os.path.join(MODEL_DIR, link_name or os.path.basename(filename))
    if os.path.exists(dest):
        print(f'  ✓ 이미 있음: {label}')
        return
    tmp_file = os.path.join(TMP_DIR, filename)
    if not os.path.exists(tmp_file):
        print(f'내려받는 중: {label}')
        download(label, repo_id=repo, filename=filename, local_dir=TMP_DIR)
    safe_symlink(tmp_file, dest)
    print(f'  ✓ 완료: {label} (바로가기 연결됨)')


# ── [행글라이터] 내려받기 전에 저장 공간부터 확인합니다 ──────
# 45GB를 20분 넘게 받다가 마지막에 "용량 부족"으로 실패하면
# 그 시간이 전부 날아갑니다. 그래서 시작하기 전에 먼저 확인합니다.
_need_gb = 50
_t, _u, _free = shutil.disk_usage(TMP_DIR)
_free_gb = _free / 1024**3
print(f'저장 공간 확인: {TMP_DIR} 에 {_free_gb:.1f} GB 남음 (약 {_need_gb} GB 필요)')
if _free_gb < _need_gb:
    print()
    print(f'⛔ 공간이 부족합니다! ({_free_gb:.1f} GB 남음 / {_need_gb} GB 필요)')
    print('   ── 이렇게 해보세요 ──')
    print('   1) 위 메뉴 Run → Stop Session 으로 세션을 완전히 끕니다.')
    print('   2) 새 세션을 시작하면 저장 공간이 비워진 상태로 배정됩니다.')
    print('   3) 다른 노트북을 동시에 켜두었다면 그것도 Stop Session 해주세요.')
    raise SystemExit('저장 공간 부족 — 위 안내를 따라주세요')
print('✅ 공간 충분합니다. 내려받기를 시작합니다.')
print()

# ── (1) 영상 만드는 두뇌 (약 16.5 GB) ───────────────────────
TRANSFORMER_FILE = 'LTX-2.3-22B-distilled-1.1-Q4_K_M.gguf'
TRANSFORMER_RENAMED = 'ltx-2.3-22b-distilled-1.1-Q4_K_M.gguf'  # 프로그램이 찾는 이름 (소문자)
fetch_and_link(TRANSFORMER_FILE, REPO,
               '영상 생성 두뇌 Q4_K_M (약 16.5 GB — 제일 오래 걸립니다)',
               link_name=TRANSFORMER_RENAMED)


# ── (2) 큰 부품들 (약 14.5 GB) ──────────────────────────────
for f, label in [
    ('ltx-2.3-22b-distilled-lora-384.safetensors', 'distilled LoRA — 속도 가속기 (약 7.1 GB)'),
    ('ltx-2.3-22b_embeddings_connector.safetensors', '임베딩 커넥터 (약 3.8 GB)'),
    ('ltx-2.3-22b_text_embedding_projection.safetensors', '텍스트 프로젝션 (약 2.2 GB)'),
    ('ltx-2.3-22b_vae.safetensors', 'VAE (약 1.4 GB)'),
]:
    fetch_and_link(f, COMPANION_REPO, label)


# ── (3) 작은 부품들 (약 1.5 GB) ─────────────────────────────
for f, label in [
    ('ltx-2.3-22b_audio_vae.safetensors', '오디오 VAE (107 MB)'),
    ('ltx-2.3-22b_vocoder.safetensors', '보코더 (258 MB)'),
    ('ltx-2.3-spatial-upscaler-x2-1.1.safetensors', '화질 업스케일러 (996 MB)'),
    ('ltx-2.3-temporal-upscaler-x2-1.0.safetensors', '프레임 업스케일러 (262 MB)'),
]:
    fetch_and_link(f, COMPANION_REPO, label)


# ── (4) Gemma 텍스트 인코더 — 내 글을 이해하는 부분 (약 12.3 GB) ──
GEMMA_FOLDER = 'gemma-3-12b-it-qat-q4_0-unquantized'
GEMMA_FILES = [
    'gemma-3-12b-it-qat-q4_0-unquantized_quanto_bf16_int8.safetensors',
    'added_tokens.json', 'chat_template.json', 'config_light.json',
    'generation_config.json', 'preprocessor_config.json', 'processor_config.json',
    'special_tokens_map.json', 'tokenizer.json', 'tokenizer.model', 'tokenizer_config.json',
]
gemma_dest = os.path.join(MODEL_DIR, GEMMA_FOLDER)
gemma_tmp = os.path.join(TMP_DIR, GEMMA_FOLDER)
os.makedirs(gemma_tmp, exist_ok=True)

for gf in GEMMA_FILES:
    if not os.path.exists(os.path.join(gemma_tmp, gf)):
        print(f'내려받는 중: Gemma 텍스트 인코더 / {gf}')
        download(gf, repo_id=COMPANION_REPO, filename=f'{GEMMA_FOLDER}/{gf}', local_dir=TMP_DIR)

if safe_symlink(gemma_tmp, gemma_dest):
    print('  ✓ 완료: Gemma 텍스트 인코더 (바로가기 연결됨)')
else:
    print('  ✓ 이미 있음: Gemma 텍스트 인코더')


# ── (5) 뒷정리: 다운로드하면서 생긴 캐시를 지워 저장 공간을 확보합니다 ──
for d in [os.path.join(MODEL_DIR, '.cache'), os.path.join(TMP_DIR, '.cache'), '/kaggle/tmp/hf_home']:
    if os.path.exists(d):
        try:
            shutil.rmtree(d)
        except Exception:
            pass

# 남은 저장 공간을 보여줍니다. (남은 공간이 0이면 용량 부족입니다)
print('\n=== 남은 저장 공간 ===')
for path in ['/kaggle/working', '/kaggle/tmp']:
    if os.path.exists(path):
        t, u, f = shutil.disk_usage(path)
        print(f'  {path:16s} 남은 공간 {f/1024**3:6.1f} GB / 전체 {t/1024**3:6.1f} GB')

print('\n✅ 3단계 완료! 모델 준비가 끝났습니다.')
````

````python
%%writefile run_ltx_t2v.py
# -*- coding: utf-8 -*-
# ── [행글라이터] 기본 네거티브 프롬프트 ──────────────────────
# 영상 100편을 직접 돌려서 확정한 값입니다.
# 이걸 비워두면(원본은 n_prompt="" 였습니다) 화면에 자막·워터마크가
# 그대로 박혀서 나오는 일이 자주 생깁니다. LTX가 학습 데이터의 자막까지
# 따라 그리기 때문입니다.
# 바꾸고 싶으면 이 줄만 고치면 됩니다.
DEFAULT_NEGATIVE = "subtitles, captions, watermark, on-screen text, logo, low quality"
# ============================================================
#  LTX-2.3 Q4 영상 생성기 — 한국어판 (Kaggle T4 전용)
#  한국어 버전 제작: @hangglwriter (https://www.youtube.com/@hangglwriter)
#  원본: AIQUEST Academy
#    - Kaggle: https://www.kaggle.com/code/unlockbulk/ltx-2-3-22b-distilled-1-1-q4-t2v-kaggle
#    - 영상  : https://youtu.be/OiKtEhoqxto
#  모델: LTX-2.3 (Lightricks) / 엔진 Wan2GP (DeepBeepMeep)
# ============================================================
import gc
import os
import sys
import json
import random
import tempfile
import glob
import traceback
import numpy as np
import subprocess
import psutil
from PIL import Image

# ── Wan2GP 폴더 안으로 들어가서 작업합니다 ──────────────────
WAN2GP_DIR = os.path.abspath("Wan2GP")
sys.path.insert(0, WAN2GP_DIR)
os.chdir(WAN2GP_DIR)

# GPU 메모리를 아껴 쓰기 위한 설정입니다.
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True,max_split_size_mb:128,garbage_collection_threshold:0.5"
os.environ["TOKENIZERS_PARALLELISM"] = "false"
os.environ["CUDA_LAUNCH_BLOCKING"] = "0"

import torch

# ── GPU 확인: CPU 세션이면 모델을 읽기 전에 바로 알려주고 멈춥니다 ──
if not torch.cuda.is_available():
    print('=' * 60)
    print('⛔ GPU 를 사용할 수 없습니다! (CPU 세션에서 실행 중)')
    print('   새 노트북 파일을 올리면 Kaggle 설정이 초기화됩니다.')
    print('   → 오른쪽 Settings → Accelerator → GPU T4 x2 선택')
    print('   → 같은 곳에서 Internet 도 On 인지 확인')
    print('   → 그 뒤 Run → Restart & Run All 로 처음부터 다시 실행하세요.')
    print('=' * 60)
    sys.exit(1)
import gradio as gr
from shared.utils.audio_video import save_video

# ── GPU 확인 ────────────────────────────────────────────────
# 이 노트북은 T4(sm_75) 전용입니다. P100(sm_60) 관련 우회 코드는 모두 제거했습니다.
_GPU_SM = torch.cuda.get_device_capability() if torch.cuda.is_available() else (0, 0)
print(f"GPU: {torch.cuda.get_device_name()}")
print(f"연산 능력(Compute Capability): sm_{_GPU_SM[0]}{_GPU_SM[1]}")
print(f"VRAM: {torch.cuda.get_device_properties(0).total_memory / 1024**3:.1f} GB")
if _GPU_SM[0] == 6:
    print("⛔ P100 이 감지되었습니다! 이 노트북은 T4 전용이라 여기서 오류가 날 가능성이 높습니다.")
    print("   → Settings → Accelerator → GPU T4 x2 로 바꾼 뒤 Restart & Run All 하세요.")
ram = psutil.virtual_memory()
print(f"RAM: 전체 {ram.total / 1024**3:.1f} GB / 사용 가능 {ram.available / 1024**3:.1f} GB")
sys.stdout.flush()

# T4 에서 가장 안정적인 attention 방식을 고릅니다.
torch.backends.cuda.enable_flash_sdp(False)
torch.backends.cuda.enable_mem_efficient_sdp(True)
torch.backends.cuda.enable_math_sdp(True)


def _register_gguf_handler():
    """GGUF(압축된 모델 형식) 를 읽을 수 있도록 등록합니다.

    Wan2GP 를 통째로 실행하면 알아서 등록되지만,
    우리처럼 스크립트만 따로 돌릴 때는 직접 등록해줘야 합니다.
    """
    import shared.qtypes.gguf
    print("  [GGUF] ✅ GGUF 읽기 기능 등록 완료")


def _patch_ltx2_config_loading():
    """GGUF 파일에서 설정값을 못 읽을 때 대비책을 마련해 둡니다."""
    import models.ltx2.ltx2 as ltx2_mod

    def _patched(path, fallback_config_path=None):
        from mmgp import quant_router
        if isinstance(path, (list, tuple)):
            path = path[0] if path else ""
        if not path:
            return {}
        try:
            _, metadata = quant_router.load_metadata_state_dict(path)
            if metadata:
                config_raw = metadata.get("config")
                if config_raw:
                    config = ltx2_mod._normalize_config(config_raw)
                    if config:
                        return config
        except Exception as e:
            print(f"  [GGUF] 설정 읽기 실패({type(e).__name__}) — 예비 설정을 씁니다")
        if fallback_config_path and os.path.isfile(fallback_config_path):
            try:
                with open(fallback_config_path, "r", encoding="utf-8") as f:
                    config = ltx2_mod._normalize_config(json.load(f))
                    if config:
                        print(f"  [GGUF] ✅ 예비 설정 사용: {os.path.basename(fallback_config_path)}")
                        return config
            except Exception:
                pass
        return {}

    ltx2_mod._load_config_from_checkpoint = _patched
    print("  [GGUF] ✅ 설정 읽기 대비책 적용 완료")


# ============================================================
#  1) AI 모델을 메모리에 올리기 (3~5분 걸립니다)
# ============================================================
print("\nLTX-2.3 22B Q4 모델을 불러오는 중입니다... (3~5분 걸려요)")
sys.stdout.flush()

from mmgp import offload
from shared.utils import files_locator as fl

fl.set_checkpoints_paths(["models", "ckpts", "."])

from models.ltx2.ltx2_handler import family_handler

_register_gguf_handler()
_patch_ltx2_config_loading()

base_model_type = "ltx2_22B"
model_def = {"ltx2_pipeline": "distilled"}   # 8단계만 계산하는 빠른 버전
extra = family_handler.query_model_def(base_model_type, model_def)
model_def.update(extra)

# 글을 이해하는 부분(Gemma) 파일 찾기
gemma_folder = "models/gemma-3-12b-it-qat-q4_0-unquantized"
gemma_files = sorted(glob.glob(os.path.join(gemma_folder, "*.safetensors")))
quanto_files = [f for f in gemma_files if "quanto" in f]
text_encoder_file = quanto_files[0] if quanto_files else (gemma_files[0] if gemma_files else None)
if not text_encoder_file:
    raise FileNotFoundError(f"{gemma_folder} 에 모델 파일이 없습니다. → 3단계 셀을 다시 실행하세요.")
print(f"  텍스트 인코더: {os.path.basename(text_encoder_file)}")

# 영상을 만드는 두뇌 찾기
transformer_path = os.path.join("models", "ltx-2.3-22b-distilled-1.1-Q4_K_M.gguf")
if not os.path.isfile(transformer_path):
    raise FileNotFoundError(f"{transformer_path} 가 없습니다. → 3단계 셀을 다시 실행하세요.")
print(f"  영상 생성 모델: {os.path.basename(transformer_path)}")
sys.stdout.flush()

# T4 의 텐서 코어는 FP16 에서 65 TFLOPS 가 나옵니다. FP32(8 TFLOPS)보다 8배 빠릅니다.
MODEL_DTYPE = torch.float16
VAE_DTYPE = torch.float16

ltx2_model, pipe = family_handler.load_model(
    model_filename=transformer_path,
    model_type="ltx2_22B_distilled",
    base_model_type=base_model_type,
    model_def=model_def,
    dtype=MODEL_DTYPE,
    VAE_dtype=VAE_DTYPE,
    text_encoder_filename=text_encoder_file,
)

# 어떤 부품들이 올라왔는지 확인용으로 찍어봅니다.
print("\n--- 불러온 부품들 ---")
for name, component in pipe.items():
    if component is None:
        print(f"  {name}: 없음")
        continue
    ctype = type(component).__name__
    if hasattr(component, 'parameters'):
        try:
            p = next(component.parameters())
            print(f"  {name}: {ctype} (dtype={p.dtype})")
        except StopIteration:
            print(f"  {name}: {ctype}")
    else:
        print(f"  {name}: {ctype}")
sys.stdout.flush()

# ============================================================
#  2) 메모리 분배 설정 (mmgp Profile 4)
# ============================================================
# 모델 전체가 16GB GPU 에 안 들어가서, 필요한 부분만 그때그때 GPU 에 올렸다 내렸다 합니다.
# 아래 budgets 숫자는 각 부분에 허용할 GPU 메모리(MB)입니다.
print("\n메모리 분배를 설정하는 중...")
sys.stdout.flush()

offload.profile(
    pipe,
    profile_no=4,
    quantizeTransformer=False,
    convertWeightsFloatTo=torch.float16,
    budgets={
        # transformer 6000 이 속도/안정성이 가장 좋은 지점입니다.
        "transformer":       6000,   # 영상 만드는 두뇌
        "text_encoder":      1500,   # 글 이해하는 부분
        "video_encoder":     2000,
        "video_decoder":     3000,
        "audio_encoder":     1000,
        "audio_decoder":     1000,
        "vocoder":            500,
        "spatial_upsampler": 1500,   # 화질 올리는 부분
        "vae":               1000,
        "*":                 1000,   # 나머지
    },
)
offload.shared_state["_attention"] = "sdpa"

print("\n✅ 준비 완료! 이제 영상을 만들 수 있습니다.")
sys.stdout.flush()


# ============================================================
#  3) 화면 크기 계산
# ============================================================
def get_resolution(base_res_str, aspect_ratio_str):
    """고른 화질과 비율을 실제 가로/세로 픽셀 수로 바꿉니다."""
    base_resolutions = {"1080p": 1088, "720p": 704, "540p": 544, "480p": 480}
    ratios = {
        "16:9 가로": 16/9, "4:3 가로": 4/3,
        "1:1 정사각형": 1.0, "3:4 세로": 3/4, "9:16 세로": 9/16,
    }
    base = base_resolutions.get(base_res_str, 704)
    ratio = ratios.get(aspect_ratio_str, 16/9)
    if ratio >= 1.0:
        height = base
        width = int(base * ratio)
    else:
        width = base
        height = int(base / ratio)
    # AI 모델 특성상 가로세로가 32의 배수여야 합니다.
    return (width // 32) * 32, (height // 32) * 32


# 화면에 표시할 길이 ↔ 실제 프레임 수
DURATION_MAP = {
    "2초 (49프레임)":   49,
    "3초 (73프레임)":   73,
    "5초 (121프레임)":  121,
    "8초 (193프레임)":  193,
    "10초 (241프레임)": 241,
    "15초 (361프레임)": 361,
}


# ============================================================
#  4) 실제로 영상을 만드는 함수
# ============================================================
@torch.inference_mode()
def Video_Generation(prompt, input_image_start, input_image_end, seed, duration_dropdown,
                     resolution_dropdown, aspect_ratio_dropdown,
                     guide_scale=3.0, num_steps=8, progress=gr.Progress()):
    try:
        gc.collect(); torch.cuda.empty_cache(); torch.cuda.synchronize()
        progress(0, desc="준비하는 중...")

        if not prompt or not prompt.strip():
            return None, "❌ 프롬프트를 적어주세요. 어떤 영상을 원하는지 영어로 쓰시면 됩니다."

        frame_rate = 24.0
        num_frames = DURATION_MAP.get(duration_dropdown, 121)
        width, height = get_resolution(resolution_dropdown, aspect_ratio_dropdown)

        # ── 시드 정하기 ──
        # 같은 글 + 같은 시드 = 항상 똑같은 영상이 나옵니다.
        if seed is None or seed < 0:
            seed = random.randint(0, 2**32 - 1)
        seed = int(seed)

        # ── 올린 사진 읽기 (안 올려도 됩니다) ──
        image_start = None
        image_end = None
        if input_image_start is not None:
            image_start = Image.open(input_image_start).convert("RGB")
        if input_image_end is not None:
            image_end = Image.open(input_image_end).convert("RGB")

        if image_start is None and image_end is not None:
            return None, "❌ 끝 사진만 넣을 수는 없습니다. 시작 사진도 같이 넣어주세요."

        if image_start is None:
            mode = "글로만 만들기"
        elif image_end is not None:
            mode = "시작 사진 → 끝 사진 사이 채우기"
        else:
            mode = "시작 사진에서 출발"

        free_vram = torch.cuda.mem_get_info()[0] / 1024**3
        ram = psutil.virtual_memory()
        print(f"\n{'='*60}")
        print(f"영상 생성 시작 [{mode}]: {width}x{height}, {num_frames}프레임, 시드={seed}")
        print(f"  GPU 여유: {free_vram:.2f} GB | RAM 여유: {ram.available / 1024**3:.1f} GB")
        print(f"  프롬프트 반영 강도: {guide_scale}")
        print(f"  프롬프트: {prompt[:120]}{'...' if len(prompt) > 120 else ''}")
        print(f"{'='*60}")
        sys.stdout.flush()

        # 한 번에 처리할 조각 크기. 256 이 메모리 부족을 가장 잘 피합니다.
        vae_tile_size = 256

        total_steps = [8]
        current_step = [0]
        current_pass = [1]

        def cb(step, latent, is_start, override_num_inference_steps=None, pass_no=None, **kwargs):
            """진행률 막대를 갱신합니다. 2단계로 나눠서 만듭니다(밑그림 → 화질 개선)."""
            if is_start:
                if override_num_inference_steps is not None:
                    total_steps[0] = override_num_inference_steps
                if pass_no is not None:
                    current_pass[0] = pass_no
                current_step[0] = 0
                return
            current_step[0] += 1
            stage_name = ("1단계 (밑그림)" if current_pass[0] == 1
                          else ("2단계 (화질 개선)" if current_pass[0] == 2 else "생성 중"))
            free_v = torch.cuda.mem_get_info()[0] / 1024**3
            print(f"  [{stage_name}] {current_step[0]}/{total_steps[0]} | GPU 여유: {free_v:.2f} GB")
            sys.stdout.flush()
            frac = current_step[0] / max(total_steps[0], 1)
            frac = (0.73 + 0.22 * frac) if current_pass[0] == 2 else (frac * 0.73)
            progress(min(frac, 0.95), desc=f"{stage_name}: {current_step[0]}/{total_steps[0]}")

        _stage_labels = {
            "VAE Encoding": ("🎞️ 올린 사진을 읽는 중...", 0.05),
            "VAE Decoding": ("🎬 결과를 영상으로 바꾸는 중...", 0.88),
            "Upsampling":   ("🔭 화질을 올리는 중...", 0.83),
        }

        import time as _time
        _t = [_time.time()]

        def set_progress_status(status: str):
            dt = _time.time() - _t[0]
            _t[0] = _time.time()
            label, frac = _stage_labels.get(status, (f"⏳ {status}...", 0.85))
            print(f"  [{status}] {label}  (+{dt:.1f}초)")
            sys.stdout.flush()
            progress(frac, desc=label)

        gen_kwargs = dict(
            input_prompt=prompt,
            image_start=image_start,
            height=height,
            width=width,
            frame_num=num_frames,
            fps=frame_rate,
            seed=seed,
            callback=cb,
            VAE_tile_size=vae_tile_size,
            input_video_strength=1.0,
            denoising_strength=1.0,
            guide_scale=float(guide_scale),
            sampling_steps=int(num_steps),
            guide_phases=2,
            n_prompt=DEFAULT_NEGATIVE,
            enhance_prompt=False,
            video_prompt_type="",
            audio_prompt_type="",
            set_progress_status=set_progress_status,
        )
        if image_end is not None:
            gen_kwargs["image_end"] = image_end

        print("  영상 생성을 시작합니다... (밑그림 → 화질 개선 → 영상으로 변환)")
        sys.stdout.flush()
        _t[0] = _time.time()
        _start = _time.time()

        # 여기가 진짜 AI 가 영상을 만드는 부분입니다.
        result = ltx2_model.generate(**gen_kwargs)
        _total_sec = _time.time() - _start

        progress(0.97, desc="✅ 영상 저장 중...")
        print(f"  [전체 소요] {_total_sec:.1f}초")
        sys.stdout.flush()

        if result is None:
            return None, "❌ 영상 생성에 실패했거나 중간에 취소되었습니다."

        # ── 결과 꺼내기 ──
        if isinstance(result, dict):
            video_tensor = result.get("x")
            audio_data = result.get("audio")
            audio_sr = result.get("audio_sampling_rate", 24000)
        elif isinstance(result, tuple):
            video_tensor = result[0]
            audio_data = result[1] if len(result) > 1 else None
            audio_sr = result[2] if len(result) > 2 else 24000
        else:
            video_tensor = result
            audio_data, audio_sr = None, 24000

        if video_tensor is None or not torch.is_tensor(video_tensor):
            return None, f"❌ 영상 데이터가 없습니다. (받은 것: {type(video_tensor)})"

        video_tensor = video_tensor.cpu()
        gc.collect(); torch.cuda.empty_cache()

        # ── mp4 파일로 저장 ──
        out_path = tempfile.mktemp(suffix=".mp4")
        video_for_save = video_tensor.unsqueeze(0).float() / 127.5 - 1.0
        save_video(tensor=video_for_save, save_file=out_path, fps=frame_rate,
                   normalize=True, value_range=(-1, 1))

        # ── 모델이 소리도 만들었다면 영상에 합칩니다 ──
        if audio_data is not None:
            try:
                import soundfile as sf
                audio_tmp = tempfile.mktemp(suffix=".wav")
                if isinstance(audio_data, np.ndarray):
                    audio_np = audio_data
                    if audio_np.ndim == 2 and audio_np.shape[0] <= 2:
                        audio_np = audio_np.T
                    sf.write(audio_tmp, audio_np, int(audio_sr or 24000))
                elif torch.is_tensor(audio_data):
                    import torchaudio
                    cpu_audio = audio_data.cpu().float()
                    if cpu_audio.dim() == 1:
                        cpu_audio = cpu_audio.unsqueeze(0)
                    if cpu_audio.dim() == 3:
                        cpu_audio = cpu_audio.squeeze(0)
                    torchaudio.save(audio_tmp, cpu_audio, int(audio_sr or 24000))
                else:
                    raise ValueError(f"알 수 없는 오디오 형식: {type(audio_data)}")

                final_path = out_path.replace(".mp4", "_with_audio.mp4")
                subprocess.run([
                    "ffmpeg", "-y", "-i", out_path, "-i", audio_tmp,
                    "-c:v", "copy", "-c:a", "aac", "-b:a", "192k",
                    "-shortest", final_path
                ], check=True, capture_output=True)
                if os.path.exists(final_path) and os.path.getsize(final_path) > 0:
                    out_path = final_path
                    print("  ✅ 소리를 영상에 합쳤습니다")
            except Exception as e:
                print(f"  ⚠️ 소리 합치기 실패 (영상은 정상): {e}")

        del video_tensor, video_for_save
        gc.collect(); torch.cuda.empty_cache()
        progress(1.0, desc="완성!")
        return out_path, (f"✅ 완성! ({_total_sec:.0f}초 걸림) | 시드: {seed} | "
                          f"{width}x{height} | {num_frames}프레임\n"
                          f"   영상 오른쪽 아래 ⬇️ 버튼으로 꼭 내려받으세요!")

    except Exception as e:
        # 오류가 나면 아래 로그에 자세한 내용이 찍히고, 화면에는 요약이 뜹니다.
        traceback.print_exc()
        gc.collect(); torch.cuda.empty_cache()
        msg = str(e)
        if "out of memory" in msg.lower():
            return None, ("❌ GPU 메모리가 부족합니다.\n"
                          "   → 화질을 540p 나 480p 로, 길이를 3초나 5초로 줄여보세요.")
        return None, f"❌ 오류: {msg}"


# ============================================================
#  5) 화면(UI) 만들기
# ============================================================
CSS = """@import url('https://fonts.googleapis.com/css2?family=Noto+Sans+KR:wght@400;600;700&family=Inter:wght@400;600;700&display=swap');
* { font-family: 'Inter', 'Noto Sans KR', sans-serif !important; }
.gradio-container { max-width: 1000px !important; margin: auto !important; }
.brand-header { text-align: center; background: linear-gradient(135deg, #6a11cb 0%, #2575fc 100%); padding: 28px; border-radius: 15px; margin-bottom: 20px; box-shadow: 0 10px 25px rgba(102,126,234,0.3); }
.brand-title { color: white; font-size: 1.9em; font-weight: 700; margin: 0 0 6px 0; }
.brand-subtitle { color: rgba(255,255,255,0.9); font-size: 1em; margin-bottom: 18px; }
.social-buttons { display: flex; justify-content: center; }
.youtube-btn { padding: 14px 36px; border-radius: 10px; font-weight: 700; font-size: 17px; text-decoration: none; display: inline-block; color: white !important; transition: all 0.3s; box-shadow: 0 4px 14px rgba(0,0,0,0.25); background: linear-gradient(135deg, #FF0000 0%, #CC0000 100%); }
.youtube-btn:hover { transform: translateY(-2px); box-shadow: 0 8px 20px rgba(255,0,0,0.4); }
button.primary { background: linear-gradient(135deg, #6a11cb 0%, #2575fc 100%) !important; color: white !important; font-weight: 600 !important; border-radius: 12px !important; }
#stop-btn { background: linear-gradient(135deg, #ef4444 0%, #b91c1c 100%) !important; color: white !important; font-weight: 600 !important; border-radius: 12px !important; }
#clear-btn { background: linear-gradient(135deg, #6b7280 0%, #374151 100%) !important; color: white !important; font-weight: 600 !important; border-radius: 12px !important; }
.footer { text-align: center; padding: 20px; margin-top: 30px; border-top: 2px solid #e5e7eb; color: #6b7280; }
.footer a { color: #6a11cb; text-decoration: none; }
.credit { font-size: 12px; color: #9ca3af; line-height: 1.7; }
"""

with gr.Blocks(css=CSS, theme=gr.themes.Soft(), title="LTX-2.3 Q4 영상 생성기 | @hangglwriter") as demo:

    gr.HTML(
        '<div class="brand-header">'
        '<div class="brand-title">🎬 글로 영상을 만드는 AI</div>'
        '<div class="brand-subtitle">LTX-2.3 Q4 · Kaggle 무료 GPU로 돌리는 한국어판</div>'
        '<div class="social-buttons">'
        '<a href="https://www.youtube.com/@hangglwriter?sub_confirmation=1" target="_blank" class="youtube-btn">'
        '▶️ 유튜브 구독하기 — @hangglwriter</a>'
        '</div></div>'
    )

    gr.Markdown(
        "### 📌 가장 간단한 사용법\n"
        "**만들고 싶은 영상을 영어로 적고 [🎬 영상 만들기] 를 누르세요. 그게 전부입니다.**\n\n"
        "> 💡 **프롬프트는 영어로** 써야 결과가 좋습니다. **한글 대사는 한글로** 넣어 주세요.\n"
        "> **사진을 넣으면 결과가 더 좋습니다.** 글만으로도 되지만, 사진에서 출발하는 쪽이 더 안정적입니다.\n"
        "> 처음이라면 **5초 / 480p** 로 먼저 만들어 보세요. 마음에 들면 **720p** 로 만들어보세요.\n"
        "> 해상도가 높을수록 시간이 더 걸립니다.\n"
        "> 뭘 써야 할지 모르겠다면 아래 **예시**를 눌러보세요."
    )

    with gr.Column():
        prompt = gr.Textbox(
            label="🎨 프롬프트 — 어떤 영상을 원하는지 영어로 적으세요",
            lines=3,
            placeholder="예: A majestic eagle soaring over snowy mountain peaks at golden hour, cinematic, 4K",
            info="'무엇이 어떻게 움직이는지' 와 '카메라 움직임' 을 적으면 좋습니다. 프롬프트만 넣어도 되고, 아래 창에 사진을 넣고 프롬프트를 써도 됩니다",
        )

        gr.Examples(
            label="👇 눌러서 바로 해보기 (예시 프롬프트)",
            examples=[
                ["A majestic eagle soaring over snowy mountain peaks at golden hour, cinematic, slow motion"],
                ["Cherry blossom petals falling in slow motion on a quiet Korean street, soft morning light"],
                ["A cozy cafe window on a rainy day, raindrops sliding down the glass, warm lights inside"],
                ["Aerial drone shot flying over a neon-lit city at night, smooth camera movement"],
                ["A cat stretching lazily on a sunny windowsill, gentle camera push in"],
            ],
            inputs=[prompt],
        )

        with gr.Accordion("🖼️ 사진에서 영상 만들기", open=True):
            with gr.Row():
                input_image_start = gr.Image(type="filepath", label="🎬 시작 사진 (첫 장면)")
                input_image_end = gr.Image(type="filepath", label="🎬 끝 사진 (마지막 장면 — 선택)")
            gr.Markdown(
                "| 넣는 것 | 결과 |\n"
                "|---|---|\n"
                "| **시작 사진만** | 그 사진에서 출발해 움직이는 영상을 만듭니다 |\n"
                "| **시작 + 끝 사진** | 두 사진 사이를 자연스럽게 채웁니다 |\n"
                "| **아무것도 안 넣음** | 글만으로 영상을 새로 만듭니다 |\n\n"
                "※ **끝 사진만 넣을 수는 없습니다.** 시작 사진이 있어야 합니다."
            )

        with gr.Row():
            seed = gr.Number(
                label="🎲 시드 (-1이면 매번 다른 영상)",
                value=-1, precision=0,
                info="마음에 든 영상의 시드를 넣으면 비슷하게 다시 만들 수 있어요",
            )
            duration_dropdown = gr.Dropdown(
                label="⏱️ 영상 길이",
                choices=list(DURATION_MAP.keys()),
                value="5초 (121프레임)",
                info="길수록 오래 걸리고 메모리도 많이 씁니다",
            )

        with gr.Row():
            resolution_dropdown = gr.Dropdown(
                label="📐 화질",
                choices=["1080p", "720p", "540p", "480p"],
                # [행글라이터] 기본값 720p → 480p (T4 실측 480p 약 6분 · 720p 약 16분)
                value="480p",
                info="480p 로 확인하고 720p 로 올리세요. 메모리 부족 오류가 나면 540p",
            )
            aspect_ratio_dropdown = gr.Dropdown(
                label="📏 화면 비율",
                choices=["16:9 가로", "4:3 가로", "1:1 정사각형", "3:4 세로", "9:16 세로"],
                value="16:9 가로",
                info="유튜브는 16:9, 쇼츠·릴스는 9:16",
            )

        guide_scale = gr.Slider(
            label="🎯 프롬프트 반영 강도",
            minimum=1.0, maximum=8.0, step=0.5, value=3.0,
            info="3 이 가장 좋습니다 | 4 이상 = 글에 더 충실 | 1~2 = AI 마음대로",
        )
        num_steps = gr.Slider(
            label="⚡ 생성 단계",
            minimum=2, maximum=8, step=1, value=8,
            info="6 = 조금 빠름 | 8 = 가장 좋은 화질 (기본)",
        )

        with gr.Row():
            gen_btn = gr.Button("🎬 영상 만들기", variant="primary", size="lg", elem_id="gen-btn")
            stop_btn = gr.Button("🛑 중단", variant="secondary", size="lg", elem_id="stop-btn")
            clear_btn = gr.Button("🗑️ 초기화", variant="secondary", size="lg", elem_id="clear-btn")

        video_out = gr.Video(label="🎥 완성된 영상 (오른쪽 아래 ⬇️ 로 내려받으세요)")
        status_out = gr.Textbox(label="ℹ️ 진행 상태", interactive=False, lines=2)

        gen_event = gen_btn.click(
            fn=Video_Generation,
            inputs=[prompt, input_image_start, input_image_end, seed, duration_dropdown,
                    resolution_dropdown, aspect_ratio_dropdown, guide_scale, num_steps],
            outputs=[video_out, status_out],
        )
        stop_btn.click(fn=None, cancels=[gen_event])
        clear_btn.click(
            fn=lambda: (None, None, None, "", -1, ""),
            outputs=[input_image_start, input_image_end, video_out, prompt, seed, status_out],
        )

    gr.HTML(
        '<div class="footer">'
        '<p style="font-size:16px;margin:5px 0;">🎬 한국어판 제작 · '
        '<strong><a href="https://www.youtube.com/@hangglwriter" target="_blank">@hangglwriter</a></strong></p>'
        '<p style="font-size:13px;margin:5px 0;">LTX-2.3 22B Distilled 1.1 Q4_K_M · Wan2GP · Kaggle 무료 GPU (T4)</p>'
        '<p class="credit">원본 노트북: <strong>AIQUEST Academy</strong> — '
        '<a href="https://www.kaggle.com/code/unlockbulk/ltx-2-3-22b-distilled-1-1-q4-t2v-kaggle" target="_blank">원본 Kaggle 노트북</a>'
        ' · <a href="https://youtu.be/OiKtEhoqxto" target="_blank">참고 영상</a>'
        ' · <a href="https://www.youtube.com/@AIQuestAcademy" target="_blank">@AIQuestAcademy</a><br>'
        '모델: <a href="https://huggingface.co/Lightricks" target="_blank">LTX-2.3 by Lightricks</a>'
        ' · <a href="https://github.com/DeepBeepMeep/Wan2GP" target="_blank">Wan2GP by DeepBeepMeep</a></p>'
        '</div>'
    )

# ============================================================
#  6) 화면 띄우기
# ============================================================
print("\n화면을 띄우는 중입니다... 아래에 나오는 gradio.live 주소를 클릭하세요!")
sys.stdout.flush()
# ── 예비 접속 주소 만들기 (gradio.live 가 504 등으로 먹통일 때용) ──
# gradio.live 는 무료 중계 서버라 가끔 통째로 장애가 납니다.
# 그래서 Cloudflare 의 무료 임시 터널로 '예비 주소'를 하나 더 만들어 둡니다.
# 아래 로그에 두 주소가 나오니, 열리는 쪽 아무거나 쓰면 됩니다.
def _start_backup_tunnel(port=7860):
    try:
        import re as _re
        import stat as _stat
        import threading as _th
        import subprocess as _sp
        import urllib.request as _ur
        binp = "/kaggle/working/cloudflared"
        if not os.path.exists(binp):
            print("예비 터널(cloudflared) 준비 중...")
            _ur.urlretrieve(
                "https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64",
                binp)
            os.chmod(binp, os.stat(binp).st_mode | _stat.S_IEXEC)
        proc = _sp.Popen([binp, "tunnel", "--url", f"http://127.0.0.1:{port}", "--no-autoupdate"],
                         stdout=_sp.PIPE, stderr=_sp.STDOUT, text=True)

        def _reader():
            found = False
            for line in proc.stdout:      # 계속 읽어줘야 터널이 안 막힙니다
                if not found:
                    m = _re.search(r"https://[a-z0-9-]+\.trycloudflare\.com", line)
                    if m:
                        found = True
                        print(chr(10) + "=" * 60)
                        print(f"🔗 예비 접속 주소: {m.group(0)}")
                        print("   gradio.live 주소가 안 열리면(504 등) 이 주소로 접속하세요!")
                        print("=" * 60 + chr(10), flush=True)

        _th.Thread(target=_reader, daemon=True).start()
    except Exception as e:
        print(f"(예비 터널 준비 실패 — gradio.live 주소만 사용하세요: {e})")


_start_backup_tunnel(7860)

demo.queue()
demo.launch(share=True, inline=False, debug=True, show_error=True, mcp_server=True, server_port=7860)
````

> **주간 할당량 30시간 중 12시간이 그대로 날아갑니다.**

### 멈추는 방법 — 정답은 하나뿐입니다

**✅ Stop Session — 이것만이 GPU 시간을 아낍니다**

오른쪽 사이드바 아래 **Session** 영역에서 **Stop Session** 을 누르세요.
(또는 오른쪽 위 **⋮** 메뉴 → **Stop Session**)
**작업이 끝났으면 반드시 이걸 눌러주세요.**

**❌ 정지 버튼(■)만 누르는 건 GPU 시간을 아끼지 못합니다**

셀 왼쪽의 정지 버튼은 `gradio.live` 화면만 닫습니다.
세션은 그대로 살아있어서 **GPU 시간은 계속 깎입니다.**
게다가 영상 만드는 프로그램이 통째로 종료되기 때문에,
